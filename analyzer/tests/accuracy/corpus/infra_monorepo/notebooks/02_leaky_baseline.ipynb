# Tabular baseline on the image metadata

Planted defect: the scaler is fitted on the whole feature matrix in cell 3 and
the split only happens in cell 4, so every held-out row contributed its mean
and variance to the scaling. MLV101.

```python
%matplotlib inline
import sys
sys.path.insert(0, "..")

import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
frame = pd.read_csv("../data/manifest.csv")
FEATURES = ["width", "height", "bytes", "aspect", "mean_intensity"]
features = frame[FEATURES].to_numpy(dtype="float32")
labels = (frame["label"] == "dog").astype(int).to_numpy()
```

```python
scaler = StandardScaler()
scaled = scaler.fit_transform(features)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    scaled, labels, test_size=0.25, random_state=0, stratify=labels)
```

```python
clf = LogisticRegression(max_iter=500, random_state=0)
clf.fit(x_train, y_train)
predicted = clf.predict(x_test)
print(accuracy_score(y_test, predicted))
print(roc_auc_score(y_test, clf.predict_proba(x_test)[:, 1]))
```
