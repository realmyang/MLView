```python
%matplotlib inline
import sys
sys.path.insert(0, "..")

import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
frame = pd.read_csv("../data/manifest.csv")
COLUMNS = ["width", "height", "bytes", "aspect", "mean_intensity"]
matrix = frame[COLUMNS].to_numpy(dtype="float32")
target = (frame["label"] == "cat").astype(int).to_numpy()
```

```python
minmax = MinMaxScaler()
normalised = minmax.fit_transform(matrix)
```

```python
a_train, a_test, b_train, b_test = train_test_split(
    normalised, target, test_size=0.25, random_state=3, stratify=target)
```

```python
forest = RandomForestClassifier(n_estimators=200, random_state=3)
forest.fit(a_train, b_train)
print(f1_score(b_test, forest.predict(a_test)))
```
