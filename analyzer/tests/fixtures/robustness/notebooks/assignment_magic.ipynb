ROB-20. IPython's assignment magics: `x = !cmd` and `x = %magic args`.

`ingest/notebook._is_magic` asks whether the *stripped line* starts with
`%`, `!` or `?`, so a magic at assignment position is never recognised. The
line stays in the generated module, `!ls data` is a SyntaxError there, and
the notebook is skipped whole -- the training loop two cells down is never
seen.

Both spellings are here. The `%env` one is verbatim from
`wandb-examples/colabs/tables/AlphaFold_with_W&B_Align,_Fold,_Log.ipynb`
cell 59, one of the nine notebooks (of 876 in the pinned public clones)
that still fail conversion.

```
files = !ls data
print(len(files))
```

```
PATH = %env PATH
```

```
import torch
import torch.nn as nn
from torch.utils.data import DataLoader

ds = None
model = nn.Linear(16, 3)
opt = torch.optim.Adam(model.parameters())
crit = nn.CrossEntropyLoss()
loader = DataLoader(ds, batch_size=8, shuffle=True)
for xb, yb in loader:
    opt.zero_grad()
    loss = crit(model(xb), yb)
    loss.backward()
    opt.step()
```
