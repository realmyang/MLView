ROB-19. A line magic whose argument list continues onto the next line.

ROB-05 taught the converter about a shell escape continued with a
backslash. This is the other continuation Python has, and the commoner one
in a timing cell: the magic's call simply does not fit on one line, so it
runs on inside an open bracket. The magic line becomes
`pass  # mlview: magic` and the continuation is left as Python, where it is
an IndentationError -- and one IndentationError costs the whole notebook,
because every cell is concatenated into one generated module.

Found by converting the 876 notebooks in the pinned public clones: 867
parse, and this is one of the nine that do not --
`PythonDataScienceHandbook/notebooks/03.12-Performance-Eval-and-Query.ipynb`
cell 1, verbatim below.

```
import numpy as np
rng = np.random.default_rng(42)
x = rng.random(1000)
y = rng.random(1000)
```

```
%timeit np.fromiter((xi + yi for xi, yi in zip(x, y)),
                    dtype=x.dtype, count=len(x))
```

```
import torch
import torch.nn as nn
from torch.utils.data import DataLoader

ds = None
model = nn.Linear(16, 3)
opt = torch.optim.Adam(model.parameters())
crit = nn.CrossEntropyLoss()
loader = DataLoader(ds, batch_size=8, shuffle=True)
for xb, yb in loader:
    opt.zero_grad()
    loss = crit(model(xb), yb)
    loss.backward()
    opt.step()
```
